# STEP 03. 수집 데이터 명세

## 작업 계획

**왜 크롤링 전에 데이터 컬럼을 먼저 정의하는가**

실제로 웹사이트에 요청을 보내고 정보를 긁어오는 코드를 작성하기 전에, "어떤 항목을 어떤 이름의 컬럼으로 저장할 것인지"를 먼저 정해야 합니다. 컬럼을 먼저 정의하지 않고 크롤링부터 시작하면, 나중에 전처리·분석·필터링 코드를 작성할 때마다 컬럼 이름과 구조가 계속 바뀌어 앞의 작업을 다시 고쳐야 하는 일이 생깁니다.

**DataFrame의 한 행이 의미하는 것**

이 프로젝트에서 다루는 pandas `DataFrame`은 엑셀 표와 비슷한 구조입니다. 그리고 이 표의 **한 행(row)은 항상 "채용공고 한 건"**을 의미합니다. 예를 들어 3개의 채용공고를 수집했다면, DataFrame은 3개의 행을 가지게 됩니다.

**이번 단계에서 하지 않는 것**

이번 STEP에서는 `requests`, `BeautifulSoup` 등을 이용한 실제 웹 요청이나 크롤링을 전혀 하지 않습니다. 외부 사이트에 접속하지도 않고, API를 호출하지도 않습니다. 오직 "앞으로 만들 표의 설계도"만 코드로 확인합니다.

**어떤 컬럼을 사용할 것인가**

| 컬럼 | 의미 |
|------|------|
| company_name | 회사명 |
| job_title | 채용공고 제목 |
| career | 경력 조건 |
| location | 근무 지역 |
| posted_date | 등록일 |
| closing_date | 마감일 |
| job_url | 공고 URL |
| search_keyword | 어떤 검색어로 발견했는지 |

위 9개 컬럼만 가진 빈 DataFrame을 만들고, 컬럼 수와 shape을 눈으로 확인할 수 있으면 성공입니다.

**중요한 점 — 처음부터 여러 페이지를 크롤링하지 않는다**

이 프로젝트는 항상 아래 순서로 확인합니다.

한 번에 전체 목록을 긁어오지 않고, 소량으로 먼저 성공시킨 뒤 다음 STEP에서 점진적으로 범위를 넓힙니다.

**HTML 구조를 먼저 확인해야 하는 이유 — 그리고 실제로 겪은 문제**

웹페이지에서 눈으로 보이는 화면과 실제 HTML 코드 구조는 항상 단순하게 1:1로 대응하지 않습니다. 이번 STEP에서 바로 그런 경우를 실제로 겪었습니다.

처음에는 회사명, 공고 제목, 지역, 경력 정보가 들어 있는 `data-sentry-component="CardJob"` 블록만 파싱하면 충분하다고 생각했습니다. 그런데 그 블록 안에는 **등록일/마감일에 해당하는 텍스트("09/21(월) 등록" 같은 문구)가 전혀 없었습니다.** 처음에는 "이 페이지가 목록 화면에서는 날짜 정보를 아예 안 주는구나"라고 잘못 판단했었습니다.

원인을 다시 조사해보니, 이 사이트는 Next.js라는 프레임워크로 만들어져 있고, 화면에 보이는 "09/21(월) 등록" 같은 문자열은 **브라우저에서 자바스크립트가 실행된 뒤에야** 화면에 그려지는 값이었습니다. `requests`는 자바스크립트를 실행하지 않고 서버가 보내주는 원본 HTML만 받아오기 때문에, 그 완성된 문구 자체는 원본 HTML에 없는 것이 맞았습니다.

하지만 페이지 안의 `<script>` 태그를 다시 살펴보니, 화면을 그리는 데 필요한 **원본 데이터(JSON)가 그 스크립트 안에 통째로 들어 있었습니다.** 그 JSON 안에 각 공고마다 `"applicationPeriod": {"start": "...", "end": "..."}` 같은 값이 있었고, 이 값이 바로 화면에 "등록"/"마감"으로 표시되는 날짜의 원본이었습니다. (예: `start`가 `2026-09-21T00:00:00+09:00`이면 화면에는 "09/21(월) 등록"으로 표시됩니다.)

즉, **문제는 "날짜 정보가 없다"가 아니라, "화면에 보이는 위치(카드 블록)를 파싱하는 방식으로는 날짜를 찾을 수 없었다"**였습니다. 날짜 값은 같은 페이지, 같은 응답 안에 이미 들어 있었고, 위치만 다를 뿐이었습니다.

**이번 STEP에서 하지 않는 것**

- 여러 페이지 반복
- 여러 검색어 반복
- 상세 페이지 순회 (상세 페이지에 대한 추가 HTTP 요청 없음 — 날짜는 이미 받은 검색 결과 페이지 응답 안에서 찾음)
- CSV/JSON 저장
- history 비교, 중복 제거, 신규 공고 판별
- 통계 분석, Gemini 호출

**성공 기준**

- 채용공고 요소(`CardJob` 블록)에서 회사명/제목/지역/경력/URL을 찾을 수 있음
- 같은 페이지 응답 안의 `<script>` 데이터에서 등록일(`posted_date`)/마감일(`closing_date`)의 원본 값(ISO 8601 날짜 문자열)을 찾을 수 있음
- 최대 5건 정도의 공고 정보를 출력할 수 있음
- 회사명, 공고 제목, 그리고 등록일/마감일까지 실제 브라우저 화면과 비교 가능
- 반복 수집은 현재 페이지(1페이지)만 대상으로 제한, 추가 네트워크 요청 없음

````python
import requests
from bs4 import BeautifulSoup
from datetime import datetime
import re
import json

SEARCH_KEYWORD = "ax"
SEARCH_URL = "https://www.jobkorea.co.kr/Search/?stext=ax&tabType=recruit"

response = requests.get(
    SEARCH_URL,
    timeout=10,
    headers={"User-Agent": "Mozilla/5.0"},
)
response.raise_for_status()

html_text = response.text
soup = BeautifulSoup(html_text, "html.parser")


def extract_job_data_from_scripts(html_text):
    """페이지 안의 <script> 태그에 들어있는 원본 채용공고 데이터(JSON)를 찾아 리스트로 반환한다.

    이 사이트는 화면에 보이는 등록일/마감일 문구를 자바스크립트로 나중에 그리기 때문에,
    requests로 받은 HTML의 눈에 보이는 부분(카드 블록)에는 날짜 텍스트가 없다.
    대신 <script>self.__next_f.push([1, "..."])</script> 안에 화면을 그리는 데 필요한
    원본 데이터가 JSON 형태로 통째로 들어 있고, 그 안에 등록일/마감일(applicationPeriod)이 있다.
    """
    pattern = re.compile(r'self\.__next_f\.push\(\[1,"((?:[^"\\]|\\.)*)"\]\)')

    for match in pattern.finditer(html_text):
        raw = match.group(1)
        if "applicationPeriod" not in raw:
            continue

        # 자바스크립트 문자열 이스케이프(\", \n 등)를 실제 문자로 풀어준다.
        decoded = json.loads('"' + raw + '"')

        # 맨 앞의 "10:" 같은 청크 번호를 떼어내고 JSON으로 해석한다.
        payload = re.sub(r"^\d+:", "", decoded, count=1)
        try:
            parsed = json.loads(payload)
        except json.JSONDecodeError:
            continue

        jobs_found = _find_job_list(parsed)
        if jobs_found:
            return jobs_found

    return []


def _find_job_list(obj):
    """중첩된 dict/list 구조 안에서 채용공고 목록(content)을 재귀적으로 찾는다."""
    if isinstance(obj, dict):
        content = obj.get("content")
        if (
            isinstance(content, list)
            and content
            and isinstance(content[0], dict)
            and "applicationPeriod" in content[0]
        ):
            return content
        for value in obj.values():
            found = _find_job_list(value)
            if found:
                return found
    elif isinstance(obj, list):
        for item in obj:
            found = _find_job_list(item)
            if found:
                return found
    return None


job_data_list = extract_job_data_from_scripts(html_text)
print("스크립트에서 찾은 원본 공고 데이터 수:", len(job_data_list))

# 공고 번호(id) 기준으로 등록일/마감일을 바로 찾을 수 있도록 딕셔너리로 변환
job_data_by_id = {job["id"]: job for job in job_data_list}

# 실제 HTML을 먼저 확인해서 찾은, 채용공고 1건에 해당하는 반복 블록
cards = soup.select('div[data-sentry-component="CardJob"]')
print("찾은 채용공고 후보 수:", len(cards))

jobs = []
for card in cards[:5]:
    title_link = card.select_one('a[data-sentry-component="Title"]')
    title_span = card.select_one('a[data-sentry-component="Title"] span')
    company_span = card.select_one("span.text-typo-b2-16")
    location_span = card.select_one('div[data-sentry-component="GrayChip"] span.text-typo-b4-14')
    career_span = card.select_one("span.flex-shrink-0.text-gray700.text-typo-c1-13")

    job_url = title_link["href"] if title_link and title_link.has_attr("href") else None

    # job_url 안의 숫자(GI_Read/50032017)가 스크립트 데이터의 id와 같다.
    job_id_match = re.search(r"GI_Read/(\d+)", job_url) if job_url else None
    job_id = job_id_match.group(1) if job_id_match else None
    matched = job_data_by_id.get(job_id)

    if matched and matched.get("applicationPeriod"):
        posted_date = matched["applicationPeriod"].get("start")
        closing_date = matched["applicationPeriod"].get("end")
    else:
        posted_date = None
        closing_date = None

    job = {
        "company_name": company_span.get_text(strip=True) if company_span else None,
        "job_title": title_span.get_text(strip=True) if title_span else None,
        "career": career_span.get_text(strip=True) if career_span else None,
        "location": location_span.get_text(strip=True) if location_span else None,
        "posted_date": posted_date,
        "closing_date": closing_date,
        "job_url": job_url,
        "search_keyword": SEARCH_KEYWORD,
        "collected_at": datetime.now().isoformat(timespec="seconds"),
    }
    jobs.append(job)

print("수집 건수:", len(jobs))

for job in jobs:
    print(job)
````

이번 STEP에서는 Gemini API를 호출하지 않습니다. `new_jobs`의 `job_title`을 `lower()`로 소문자로 바꿔 키워드와 비교합니다. 제목에 키워드가 **하나라도 포함되면** 관련 공고 후보로 판단합니다. 회사명·경력·지역은 결과를 읽을 때 참고하며 필터 조건으로 사용하지 않습니다.

이 작업은 **1차 후보 추리기**이며 최종 직무 판단이 아닙니다. 현재 5건은 이미 검색어 `ax`로 수집했으므로 대부분 또는 전부가 포함되어도 이상하지 않습니다. 목표는 좋은 필터 모델을 만드는 것이 아니라 규칙 기반으로 공고를 골라내는 구조를 이해하는 것입니다. 통계는 현재 5건 기준이며 전체 채용시장의 비율로 해석하지 않습니다.

**실행 및 확인할 내용 / 성공 기준**

- 키워드 목록을 정의하고 제목에서 대소문자를 무시하여 검색합니다.
- `new_jobs`의 복사본에 `matched_keywords`를 기록합니다. 빈 목록 `[]`은 매칭된 키워드가 없다는 뜻입니다.
- 관련 후보 DataFrame `related_jobs`를 생성하고 공고 수와 비율을 출력합니다.
- 전체 공고의 생략 없는 제목과 매칭 키워드를 읽어 포함 이유를 확인하고, 제외된 공고도 확인합니다.

**재실행 순서**: 커널을 재시작했다면 STEP 05 → STEP 06 → STEP 07 → STEP 08 → STEP 09 → STEP 10 → STEP 11 순서로 실행합니다. 이미 `new_jobs`가 메모리에 있으면 아래 Code Cell만 실행해도 됩니다.


````python
import pandas as pd

# 1. 제목에서 찾을 키워드 목록
keywords = [
    "ax", "ai", "인공지능", "데이터 분석", "생성형 ai", "생성형ai",
    "llm", "machine learning", "머신러닝", "data scientist",
    "데이터 사이언티스트", "ai engineer", "ai 엔지니어",
]

# 2. 제목을 소문자로 바꾸고, 포함된 키워드를 목록으로 반환합니다.
def find_matched_keywords(title):
    title_lower = str(title).lower()
    matched = []
    for keyword in keywords:
        if keyword.lower() in title_lower:
            matched.append(keyword)
    return matched

if "new_jobs" not in globals():
    print("new_jobs가 없습니다. STEP 09 Code Cell을 먼저 실행하세요.")
else:
    # 3. 원본은 유지하고 복사본에 매칭 결과를 추가합니다.
    filtered = new_jobs.copy()
    filtered["matched_keywords"] = filtered["job_title"].apply(find_matched_keywords)

    # 4. 매칭 키워드가 하나 이상인 행만 관련 공고 후보로 선택합니다.
    related_mask = filtered["matched_keywords"].apply(len) > 0
    related_jobs = filtered.loc[related_mask].copy()

    related_ratio = (
        len(related_jobs) / len(new_jobs) * 100 if len(new_jobs) > 0 else 0
    )
    print("신규 공고 수:", len(new_jobs))
    print("관련 공고 수:", len(related_jobs))
    print(f"관련 공고 비율: {related_ratio:.1f}%")
    print("사용한 키워드:", keywords)

    # 5. 긴 제목도 생략하지 않고 전체 제목과 포함/제외 이유를 확인합니다.
    result_columns = ["company_name", "job_title", "matched_keywords", "job_url"]
    with pd.option_context("display.max_colwidth", None):
        print("\n전체 신규 공고의 제목과 매칭 키워드:")
        display(filtered[result_columns])
        print("\n관련 공고 후보:")
        display(related_jobs[result_columns])
        print("\n키워드가 없어 제외된 공고:")
        if (~related_mask).any():
            display(filtered.loc[~related_mask, result_columns])
        else:
            print("없음")

    print("\n키워드별 매칭 공고 수 (한 공고가 여러 키워드에 집계될 수 있음):")
    print(related_jobs["matched_keywords"].explode().value_counts())
````

````python
import os
from pathlib import Path

from dotenv import load_dotenv
from openai import OpenAI

# 1. 현재 작업 디렉터리를 확인하고 프로젝트 루트를 계산합니다.
#    (경로를 추측하지 않고, 실제 위치를 보고 판단합니다. os.chdir()는 사용하지 않습니다.)
cwd = Path.cwd()
print("현재 작업 디렉터리:", cwd)

if cwd.name == "notebooks":
    project_root = cwd.parent
else:
    project_root = cwd

env_path = project_root / ".env"
print("프로젝트 루트:", project_root)
print(".env 경로:", env_path)

# 2. .env 파일을 읽어 환경변수로 올립니다.
load_dotenv(env_path)

# 3. 환경변수에서 값을 읽습니다. (API Key 값 자체는 절대 출력하지 않습니다)
api_key = os.getenv("OPENAI_API_KEY")
model_name = os.getenv("OPENAI_MODEL")

print(".env 파일 존재 여부:", env_path.exists())
print("OPENAI_API_KEY 존재 여부:", bool(api_key))
print("OPENAI_MODEL:", model_name)

# 4. 사전 조건을 순서대로 확인합니다.
if not env_path.exists():
    print("\n[중단] .env 파일을 찾을 수 없습니다.")
    print(f"다음 위치에 .env 파일을 만드세요: {env_path}")
elif not api_key:
    print("\n[중단] OPENAI_API_KEY가 설정되지 않았습니다.")
    print(".env 파일에 OPENAI_API_KEY=발급받은_키 형식으로 입력하세요.")
elif not model_name:
    print("\n[중단] OPENAI_MODEL이 설정되지 않았습니다.")
    print(".env 파일에 OPENAI_MODEL=사용할_모델명 형식으로 입력하세요.")
elif "related_jobs" not in globals():
    print("\n[중단] related_jobs 변수가 없습니다.")
    print("Notebook을 재시작했다면 STEP 05부터 STEP 11까지 순서대로 실행하세요.")
elif related_jobs.empty:
    print("\n[중단] related_jobs가 비어 있습니다. 관련 공고가 0건입니다.")
else:
    # 5. 테스트 대상: 관련 공고 중 첫 번째 1건만 사용합니다.
    sample_job = related_jobs.iloc[0]

    # 6. 연결 테스트용 최소 프롬프트를 만듭니다.
    #    (수집한 값만 사용하고, 없는 정보를 추측해서 넣지 않습니다.)
    prompt = f"""
다음 채용공고 정보를 보고 아래 두 가지에 답하세요.

1. 이 직무를 한 문장으로 요약하세요.
2. 이 공고가 AX와 관련된 이유를 한 문장으로 설명하세요.

회사명: {sample_job["company_name"]}
공고 제목: {sample_job["job_title"]}
경력: {sample_job["career"]}
지역: {sample_job["location"]}
등록일: {sample_job["posted_date"]}
마감일: {sample_job["closing_date"]}
"""

    print("\n=== 테스트 대상 공고 ===")
    print("회사:", sample_job["company_name"])
    print("공고:", sample_job["job_title"])
    print("모델:", model_name)

    # 7. OpenAI 공식 SDK의 Client를 만듭니다. (.env에서 읽은 키를 전달)
    client = OpenAI(api_key=api_key)

    # 8. Responses API로 호출합니다.
    try:
        response = client.responses.create(
            model=model_name,
            input=prompt,
        )

        print("\n=== OpenAI 응답 ===")
        print(response.output_text)
        print("\n응답 길이:", len(response.output_text))
        print("API 호출 성공")

    except Exception as e:
        print("\nOpenAI API 호출 중 오류가 발생했습니다.")
        print("오류 종류:", type(e).__name__)
        print("오류 내용:", str(e))
        print("\n오류 내용에 모델(model) 관련 문구가 있다면,")
        print(f".env 파일의 OPENAI_MODEL 값(현재: {model_name})을 사용 가능한 모델명으로 수정한 뒤 다시 실행하세요.")
````

# STEP 15. Slack 발송

## 작업 계획

**이번 단계에서 하는 일**

STEP 14에서 만든 Markdown 보고서를 Slack Webhook을 통해 실제 Slack 채널로 전송합니다.

**확인할 항목**

- `.env`에서 `SLACK_WEBHOOK_URL`을 읽을 수 있는가
- 보고서 파일(`reports/weekly_ax_jobs_*.md`)이 존재하는가
- Slack 요청이 성공하는가
- HTTP 상태 코드가 정상인가
- 실제 Slack 채널에 메시지가 도착하는가
- 한글이 깨지지 않는가
- 링크가 정상적으로 표시되는가

**핵심 원칙**

> **코드에서 성공 메시지가 나왔다고 해서 끝이 아니라, 실제 Slack 채널에서 사람이 메시지를 확인해야 한다.**

HTTP 요청이 성공(예: 200 OK)했다는 것은 "Slack 서버가 요청을 잘 받았다"는 뜻이지, "채널에 메시지가 정상적으로 표시되었다"는 뜻까지 보장하지는 않습니다. 그래서 이번 STEP의 진짜 완료 조건은 코드의 출력이 아니라, **사람이 Slack 앱/웹에서 채널을 열어 직접 확인하는 것**입니다.

**이번 STEP의 범위**

이번 STEP에서는 **Slack만** 구현합니다. Gmail 발송은 STEP 16에서 별도로 진행합니다.

**흐름**

```
Markdown Report
    ↓
Slack Webhook
    ↓
HTTP 응답 확인
    ↓
Slack 채널에서 메시지 직접 확인
```

**Slack Webhook URL을 다루는 원칙**

- `SLACK_WEBHOOK_URL`은 `.env`에서만 읽어옵니다. Notebook 코드에 직접 써넣지 않습니다.
- 이 값 자체는 절대 화면에 출력하지 않습니다. (`bool(...)`로 존재 여부만 확인)
- STEP 14의 보고서를 **다시 계산하거나 OpenAI API를 재호출하지 않고**, 이미 저장된 Markdown 파일을 그대로 읽어서 전송합니다.

**메시지 길이 처리**

Slack 메시지에 보고서 전체를 그대로 보내면 너무 길어질 수 있습니다. 이번 STEP에서는 복잡한 메시지 분할 로직 없이, 일정 길이(3500자)를 넘으면 그 이후를 잘라내고 "[보고서 일부 생략]"이라는 안내만 덧붙이는 단순한 방식을 사용합니다.

**Slack에서 헤더가 그대로 보이는 문제 (실제 발송 후 확인된 개선 사항)**

실제로 Slack 채널에 보내본 결과, Markdown의 `#`, `##`, `###` 헤더 문법이 굵은 글씨로 바뀌지 않고 `#` 기호가 포함된 글자 그대로 표시되는 것을 확인했습니다. Slack은 일반 Markdown이 아니라 자체 서식 규칙(mrkdwn)을 사용하기 때문입니다. 그래서 이번 STEP부터는 **Slack으로 보내는 문자열만** 별도로 만들어, 줄 앞의 `# `/`## `/`### `를 Slack에서 굵게 표시되는 `*텍스트*` 형식으로 바꿉니다. STEP 14가 저장한 `reports/*.md` **원본 파일 자체는 건드리지 않고** 일반 Markdown 형식 그대로 둡니다.

**Notebook 변수는 이전 Cell 실행 결과를 메모리에 가지고 있다는 점**

이번 STEP은 이전 STEP들의 Notebook 변수(`new_jobs`, `related_jobs` 등)가 없어도 됩니다. `reports/` 폴더에 STEP 14가 만든 보고서 파일만 있으면 STEP 15 Code Cell을 단독으로 실행할 수 있습니다.

**성공 기준**

- HTTP 요청 성공
- Slack 응답 확인
- 실제 채널 메시지 도착 확인 (사람이 직접)

````python
import os
from pathlib import Path

import requests
from dotenv import load_dotenv


cwd = Path.cwd()

if cwd.name == "notebooks":
    project_root = cwd.parent
else:
    project_root = cwd

env_path = project_root / ".env"

load_dotenv(env_path)

slack_webhook_url = os.getenv("SLACK_WEBHOOK_URL")

reports_dir = project_root / "reports"

report_files = sorted(
    reports_dir.glob("weekly_ax_jobs_*.md"),
    key=lambda p: p.stat().st_mtime,
    reverse=True,
) if reports_dir.exists() else []

print("SLACK_WEBHOOK_URL 존재 여부:", bool(slack_webhook_url))
print("reports 폴더 존재 여부:", reports_dir.exists())
print("찾은 보고서 파일 수:", len(report_files))

if not slack_webhook_url:
    print("\nSLACK_WEBHOOK_URL이 없습니다. .env를 확인하세요.")

elif not report_files:
    print("\n전송할 Markdown 보고서가 없습니다. STEP 14를 먼저 실행하세요.")

else:
    report_path = report_files[0]

    report_text = report_path.read_text(
        encoding="utf-8"
    )

    print("\n전송할 보고서:", report_path.name)
    print("보고서 길이:", len(report_text))

    # Slack Incoming Webhook은 "# / ## / ###" 같은 Markdown 헤더 문법을
    # 굵은 글씨로 자동 변환해주지 않고 문자 그대로 보여준다.
    # 그래서 원본 파일(report_text)은 그대로 두고,
    # Slack으로 보낼 문자열만 줄 단위로 간단히 변환한다.
    slack_lines = []
    for line in report_text.splitlines():
        if line.startswith("### "):
            slack_lines.append(f"*{line[4:]}*")
        elif line.startswith("## "):
            slack_lines.append(f"*{line[3:]}*")
        elif line.startswith("# "):
            slack_lines.append(f"*{line[2:]}*")
        else:
            slack_lines.append(line)
    slack_text = "\n".join(slack_lines)

    message_text = slack_text

    # 지나치게 긴 경우 초보자 실습용 단순 제한
    max_length = 3500

    if len(message_text) > max_length:
        message_text = (
            message_text[:max_length]
            + "\n\n[보고서 일부 생략]"
        )

    print("실제 전송 길이:", len(message_text))

    payload = {
        "text": message_text
    }

    try:
        response = requests.post(
            slack_webhook_url,
            json=payload,
            timeout=10,
        )

        print("\nSlack HTTP 상태 코드:", response.status_code)
        print("Slack 응답:", response.text)

        if response.ok:
            print("Slack 전송 요청 성공")
        else:
            print("Slack 전송 요청 실패")

    except requests.RequestException as e:
        print("\nSlack 전송 중 오류가 발생했습니다.")
        print("오류 종류:", type(e).__name__)
        print("오류 내용:", str(e))
````

# STEP 16. Gmail 발송

## 작업 계획

**이번 단계에서 하는 일**

준비된 `EmailMessage`를 Gmail SMTP 서버로 **실제 전송**합니다.

```
Markdown Report

Gmail SMTP

앱 비밀번호 인증

실제 수신함 확인
```

> **경고: 이 Code Cell을 실행할 때마다 실제 메일이 1건 발송됩니다.**
> **실수로 반복 실행하지 않도록 주의하세요.** (Notebook 전체를 "Run All"로 실행하지 말고, 이 Cell만 필요할 때 직접 실행하세요.)

**확인할 항목**

- 보고서 파일이 존재하는가
- `GMAIL_ADDRESS`가 있는가
- `GMAIL_APP_PASSWORD`가 있는가

- `EmailMessage`가 정상 생성되는가
- Gmail SMTP 서버 연결/로그인이 성공하는가
- 메일 전송 요청이 성공하는가
- **실제 수신함에서 메일이 도착했는가** (Python 출력만으로는 알 수 없음)

**Gmail 발송 방식**

초보자 실습 목적상 가장 단순한 SMTP 방식을 사용합니다. Python 표준 라이브러리인 `smtplib`, `email.message.EmailMessage`만 사용하며, 새 패키지는 설치하지 않습니다.

- 서버: `smtp.gmail.com`
- 포트: `465`
- 연결 방식: `smtplib.SMTP_SSL`

**Gmail 인증 방식**

일반 Gmail 계정 비밀번호를 직접 사용하지 않습니다. Google 계정의 **앱 비밀번호(App Password)**를 사용합니다.

| 환경변수 | 의미 |
|----------|------|
| `GMAIL_ADDRESS` | 발신 Gmail 주소 |
| `GMAIL_APP_PASSWORD` | Google 앱 비밀번호 (일반 로그인 비밀번호 아님) |
| `GMAIL_TO` | 수신 이메일 주소 (`.env`의 이 값 한 곳으로만 발송) |

**보안 원칙**

Gmail 주소, 앱 비밀번호, 수신자 이메일은 Notebook에 직접 작성하지 않고 모두 `.env`에서 읽습니다. 특히 `GMAIL_APP_PASSWORD` 값은 절대 화면에 출력하지 않습니다 (`bool(...)`로 존재 여부만 확인).

**발송 범위 제한**

- 이번 테스트에서는 **정확히 1건만** 발송합니다. 반복문(`for`/`while`)으로 여러 통을 보내지 않습니다.
- 메일 제목은 기존에 생성한 `[AX 채용 동향] weekly_ax_jobs_YYYY-MM-DD` 형식을 그대로 사용합니다. 테스트라는 이유로 별도 제목으로 바꾸지 않습니다.
- 오류가 발생해도 자동으로 재시도하지 않습니다. 사람이 오류 내용을 읽고 다음 행동을 결정합니다.

**Notebook 변수는 이전 Cell 실행 결과를 메모리에 가지고 있다는 점**

이번 STEP은 `reports/` 폴더의 파일만 있으면 되므로, 이전 STEP들의 Notebook 변수가 메모리에 없어도 STEP 16 Code Cell을 단독으로 실행할 수 있습니다.

**성공 기준**

- SMTP 연결 성공
- 로그인(인증) 성공
- `send_message()` 오류 없이 완료
- Python에서 "메일 전송 요청 성공" 메시지 확인
- **실제 수신함에서 메일 도착 확인** (제목/본문/한글 표시 모두 정상)

````python
import os
import smtplib
from pathlib import Path
from email.message import EmailMessage

from dotenv import load_dotenv

# 주의: 이 Code Cell을 실행할 때마다 실제 Gmail 메일이 1건 발송됩니다.
# 실수로 반복 실행하지 않도록 주의하세요.


cwd = Path.cwd()

if cwd.name == "notebooks":
    project_root = cwd.parent
else:
    project_root = cwd

env_path = project_root / ".env"

load_dotenv(env_path)

gmail_address = os.getenv("GMAIL_ADDRESS")
gmail_app_password = os.getenv("GMAIL_APP_PASSWORD")
gmail_to = os.getenv("GMAIL_TO")

print(".env 파일 존재 여부:", env_path.exists())
print("GMAIL_ADDRESS 존재 여부:", bool(gmail_address))
print("GMAIL_APP_PASSWORD 존재 여부:", bool(gmail_app_password))
print("GMAIL_TO 존재 여부:", bool(gmail_to))

reports_dir = project_root / "reports"

report_files = sorted(
    reports_dir.glob("weekly_ax_jobs_*.md"),
    key=lambda p: p.stat().st_mtime,
    reverse=True,
) if reports_dir.exists() else []

if not report_files:
    print("전송할 보고서 파일이 없습니다.")
else:
    report_path = report_files[0]

    report_text = report_path.read_text(
        encoding="utf-8"
    )

    subject = f"[AX 채용 동향] {report_path.stem}"

    msg = EmailMessage()

    if gmail_address:
        msg["From"] = gmail_address

    if gmail_to:
        msg["To"] = gmail_to

    msg["Subject"] = subject
    msg.set_content(report_text)

    print("보고서 파일:", report_path.name)
    print("메일 제목:", subject)
    print("본문 길이:", len(report_text))
    print("EmailMessage 생성 완료")

    # 여기부터 실제 Gmail SMTP 발송입니다. 정확히 1건만 보냅니다 (반복문 없음).
    if not gmail_address or not gmail_app_password or not gmail_to:
        print(
            "\nGMAIL_ADDRESS / GMAIL_APP_PASSWORD / GMAIL_TO 중 "
            "일부가 없어 발송을 진행하지 않습니다."
        )
        print(".env 파일에 세 값을 모두 입력한 뒤 다시 실행하세요.")
    else:
        print("\n=== Gmail 발송 전 확인 ===")
        print("보고서 파일:", report_path.name)
        print("메일 제목:", subject)
        print("본문 길이:", len(report_text))
        print("GMAIL_ADDRESS 존재 여부:", bool(gmail_address))
        print("GMAIL_APP_PASSWORD 존재 여부:", bool(gmail_app_password))
        print("GMAIL_TO 존재 여부:", bool(gmail_to))

        try:
            with smtplib.SMTP_SSL(
                "smtp.gmail.com",
                465,
                timeout=20,
            ) as smtp:
                smtp.login(
                    gmail_address,
                    gmail_app_password,
                )

                smtp.send_message(msg)

            print("\n=== Gmail 전송 결과 ===")
            print("Gmail SMTP 연결 및 인증 성공")
            print("메일 전송 요청 성공")
            print("수신자:", gmail_to)
            print("수신함에서 실제 도착 여부를 확인하세요.")

        except smtplib.SMTPAuthenticationError as e:
            print("\nGmail 인증에 실패했습니다.")
            print("다음 가능성을 확인하세요.")
            print("- GMAIL_ADDRESS 오타")
            print("- 앱 비밀번호 오타 또는 공백 포함")
            print("- 일반 Gmail 로그인 비밀번호를 잘못 사용")
            print("- 발급한 앱 비밀번호가 취소됨")
            print("오류 내용:", str(e))

        except smtplib.SMTPException as e:
            print("\nGmail SMTP 처리 중 오류가 발생했습니다.")
            print("오류 종류:", type(e).__name__)
            print("오류 내용:", str(e))

        except Exception as e:
            print("\nGmail 전송 중 예상하지 못한 오류가 발생했습니다.")
            print("오류 종류:", type(e).__name__)
            print("오류 내용:", str(e))
````

```
검증된 Notebook 코드

함수 정의
```

각 함수는 `입력 → 처리 → 반환` 구조가 명확하도록, 필요한 값을 모두 **인자로 받습니다.** 함수 내부에서 `global`로 Notebook 전역 변수를 직접 참조하지 않습니다. Slack Webhook URL이나 Gmail 앱 비밀번호도 함수가 `.env`를 직접 읽지 않고, 호출하는 쪽에서 인자로 넘겨줍니다.

이번 STEP의 테스트는 STEP 09~11에서 만든 `new_jobs`, `related_jobs`, `keywords` 변수를 사용합니다. 커널을 재시작했다면 최소 STEP 05 ~ STEP 11까지 다시 실행해야 합니다. STEP 12(OpenAI)/STEP 15(Slack)/STEP 16(Gmail)은 비용이 들거나 실제로 메시지가 나가므로 재실행하지 않아도 되며, 이 경우 AI 검증 예시 없이(`ai_example=None`) 테스트합니다.

**성공 기준**

- 4개 함수가 오류 없이 정의됨
- `filter_related_jobs()` 결과가 기존 STEP 11의 `related_jobs`와 동일함 (공고 수, `job_url` 집합)
- `build_markdown_report()`가 신규 공고 수/관련 공고 수/회사별 통계/관련 공고 목록을 포함한 Markdown 문자열을 정상 생성함
- `send_slack_message()`/`send_gmail()`은 정의만 되고 실제로 호출되지 않음

````python
def filter_related_jobs(df, keywords):
    """
    df의 job_title에 keywords 중 하나 이상이 포함된 행만 반환합니다.
    (STEP 11에서 검증한 로직과 동일한 방식 — matched_keywords 컬럼 생성 방식 유지)
    """
    def find_matched_keywords(title):
        title_lower = str(title).lower()
        return [kw for kw in keywords if kw.lower() in title_lower]

    filtered = df.copy()
    filtered["matched_keywords"] = filtered["job_title"].apply(find_matched_keywords)
    related_mask = filtered["matched_keywords"].apply(len) > 0
    return filtered.loc[related_mask].copy()


def build_markdown_report(new_jobs, related_jobs, ai_example=None):
    """
    신규 공고(new_jobs)와 관련 공고(related_jobs)를 바탕으로
    주간 AX 채용 동향 Markdown 문자열을 만들어 반환합니다. (STEP 14 로직과 동일, 파일 저장은 하지 않음)

    ai_example: {"company_name": ..., "job_title": ..., "response_text": ...} 형태의 dict.
                검증된 OpenAI 응답이 없으면 None으로 둡니다.
    """
    from datetime import date

    new_count = len(new_jobs)
    related_count = len(related_jobs)
    related_ratio = (related_count / new_count * 100) if new_count > 0 else 0

    company_counts = new_jobs["company_name"].value_counts()
    location_counts = new_jobs["location"].value_counts()
    career_counts = new_jobs["career"].value_counts()

    report_date = date.today().isoformat()
    lines = []

    lines.append("# 주간 AX 채용 동향")
    lines.append("")
    lines.append(f"실행일: {report_date}")
    lines.append("")

    lines.append("## 1. 이번 주 요약")
    lines.append("")
    lines.append(f"- 신규 공고: {new_count}건")
    lines.append(f"- 관련 공고: {related_count}건")
    lines.append(f"- 관련 공고 비율: {related_ratio:.1f}%")
    lines.append("- 검색어: ax")
    lines.append("")

    lines.append("## 2. 회사별 공고 수")
    lines.append("")
    for name, count in company_counts.items():
        lines.append(f"- {name}: {count}건")
    lines.append("")

    lines.append("## 3. 지역별 공고 수")
    lines.append("")
    for name, count in location_counts.items():
        lines.append(f"- {name}: {count}건")
    lines.append("")

    lines.append("## 4. 경력 조건별 공고 수")
    lines.append("")
    for name, count in career_counts.items():
        lines.append(f"- {name}: {count}건")
    lines.append("")

    lines.append("## 5. 관련 공고 목록")
    lines.append("")
    for _, job in related_jobs.head(5).iterrows():
        lines.append(f"### {job['company_name']} — {job['job_title']}")
        lines.append("")
        lines.append(f"- 경력: {job['career']}")
        lines.append(f"- 지역: {job['location']}")
        lines.append(f"- 등록일: {job['posted_date']}")
        lines.append(f"- 마감일: {job['closing_date']}")
        lines.append(f"- 공고 링크: {job['job_url']}")
        lines.append("")

    lines.append("## 6. OpenAI 검증 예시")
    lines.append("")
    if ai_example is not None:
        lines.append(f"- 회사명: {ai_example['company_name']}")
        lines.append(f"- 공고 제목: {ai_example['job_title']}")
        lines.append("")
        lines.append("**OpenAI 응답:**")
        lines.append("")
        lines.append(ai_example["response_text"])
        lines.append("")
        lines.append("**검증 결과:** 수정 후 사용 가능")
        lines.append("")
        lines.append(
            "> OpenAI 응답에는 원본 필드에서 직접 확인할 수 없는 "
            "확장 해석이 일부 포함될 수 있으므로, 최종 보고서 사용 전 원문 확인이 필요하다."
        )
    else:
        lines.append("현재 커널에 검증된 AI 응답이 없습니다. (OpenAI API를 추가로 호출하지 않았습니다.)")
    lines.append("")

    lines.append("## 7. 데이터 기준")
    lines.append("")
    lines.append("- 검색어: ax")
    lines.append("- 수집 페이지: 1페이지")
    lines.append("- 분석 대상: 현재 신규 공고")
    lines.append("- 신규 판별 기준: job_url")
    lines.append("- 관련 공고 필터 기준: 제목의 키워드 포함 여부")
    ai_count = 1 if ai_example is not None else 0
    lines.append(f"- AI 검증 공고 수: {ai_count}건")
    lines.append("")

    lines.append("## 8. 주의사항")
    lines.append("")
    lines.append("- 현재 실습 데이터는 5건뿐이므로 시장 전체를 대표하지 않는다.")
    lines.append("- 검색어도 현재 ax 하나만 사용했다.")
    lines.append("- 실제 지원 전 원문 채용공고를 다시 확인해야 한다.")
    lines.append("- OpenAI 결과는 원본에 없는 내용을 확장 해석할 수 있다.")
    lines.append("")

    return "\n".join(lines)


def send_slack_message(webhook_url, message_text):
    """
    Slack Incoming Webhook(webhook_url)으로 message_text를 전송하고 응답을 반환합니다.
    (STEP 15 로직과 동일) webhook_url은 함수 내부에서 .env를 읽지 않고 인자로 받습니다.

    주의: 실제로 호출하면 진짜 Slack 메시지가 발송됩니다.
    이번 STEP에서는 정의만 하고 호출하지 않습니다.
    """
    import requests

    response = requests.post(
        webhook_url,
        json={"text": message_text},
        timeout=10,
    )
    return response


def send_gmail(gmail_address, gmail_app_password, gmail_to, subject, body):
    """
    Gmail SMTP(smtp.gmail.com:465)로 메일 1건을 발송합니다. (STEP 16 로직과 동일)

    주의: 실제로 호출하면 진짜 메일이 발송됩니다.
    이번 STEP에서는 정의만 하고 호출하지 않습니다.
    """
    import smtplib
    from email.message import EmailMessage

    msg = EmailMessage()
    msg["From"] = gmail_address
    msg["To"] = gmail_to
    msg["Subject"] = subject
    msg.set_content(body)

    with smtplib.SMTP_SSL("smtp.gmail.com", 465, timeout=20) as smtp:
        smtp.login(gmail_address, gmail_app_password)
        smtp.send_message(msg)

    return True


print("=== 함수 정의 확인 ===")
print("filter_related_jobs 정의됨:", callable(filter_related_jobs))
print("build_markdown_report 정의됨:", callable(build_markdown_report))
print("send_slack_message 정의됨:", callable(send_slack_message))
print("send_gmail 정의됨:", callable(send_gmail))

if "new_jobs" not in globals() or "related_jobs" not in globals() or "keywords" not in globals():
    print(
        "\nnew_jobs / related_jobs / keywords가 없습니다.\n"
        "커널을 재시작했다면 STEP 05 ~ STEP 11을 순서대로 먼저 실행하세요."
    )
else:
    print("\n=== 1. filter_related_jobs() 테스트 ===")
    related_jobs_test = filter_related_jobs(new_jobs, keywords)

    print("기존 관련 공고 수:", len(related_jobs))
    print("함수 결과 공고 수:", len(related_jobs_test))

    same_urls = set(related_jobs["job_url"]) == set(related_jobs_test["job_url"])
    print("기존 결과와 동일한가:", same_urls)

    print("\n=== 2. build_markdown_report() 테스트 ===")
    if "sample_job" in globals() and ("response_text" in globals() or "response" in globals()):
        ai_example = {
            "company_name": sample_job["company_name"],
            "job_title": sample_job["job_title"],
            "response_text": response_text if "response_text" in globals() else response.output_text,
        }
    else:
        ai_example = None

    report_text_test = build_markdown_report(new_jobs, related_jobs_test, ai_example=ai_example)

    print("보고서 문자열 길이:", len(report_text_test))
    print("\n--- 미리보기 (앞부분) ---")
    print(report_text_test[:1000])

    print("\n=== 3. Slack / Gmail 함수 실행 여부 ===")
    print("send_slack_message 실제 호출: 실행 안 함 (중복 발송 방지)")
    print("send_gmail 실제 호출: 실행 안 함 (중복 발송 방지)")
````

## 작업 계획

**이번 단계에서 하는 일**

STEP 17에서 Notebook 안에 정의하고 검증했던 4개 함수를, `src/` 폴더 아래의 Python 파일로 옮깁니다. 이렇게 하면 Notebook 밖(예: 나중에 만들 `main.py`)에서도 이 함수들을 다시 불러와 쓸 수 있습니다.

**흐름**

```
Notebook 검증 함수
    ↓
src/analyzer.py
src/reporter.py
src/notifier.py
    ↓
Notebook에서 import
    ↓
같은 결과인지 확인
```

**왜 하는가**

- Notebook 밖에서도 재사용 가능 (STEP 19의 `main.py`에서 불러 쓸 예정)
- 코드를 역할별로 분리 (분석 / 보고서 / 알림)
- 유지보수가 쉬워짐 (한 파일만 열어 그 역할의 코드만 확인)

**src 폴더 구조**

```
src/
├── __init__.py     (빈 파일 — Python이 src를 "패키지"로 인식하게 해주는 표시)
├── analyzer.py      (filter_related_jobs)
├── reporter.py      (build_markdown_report)
└── notifier.py      (send_slack_message, send_gmail)
```

`__init__.py`는 내용이 없어도 됩니다. 이 파일이 있으면 Python이 `src` 폴더를 "그냥 폴더"가 아니라 "import할 수 있는 패키지"로 인식합니다.

**역할 분리 원칙**

- `analyzer.py` → 분석만 (Slack을 보내거나 SMTP에 연결하지 않음)
- `reporter.py` → Markdown 문자열 생성만 (파일 저장이나 외부 전송을 하지 않음)
- `notifier.py` → 외부 전송만 (pandas 분석을 하지 않음)

**보안 원칙**

`src/notifier.py`는 `.env`나 환경변수를 직접 읽지 않습니다. `SLACK_WEBHOOK_URL`, `GMAIL_APP_PASSWORD` 같은 비밀값은 항상 이 모듈을 **호출하는 쪽**(예: 나중의 `main.py`)에서 인자로 전달받습니다. src 코드 어디에도 실제 API Key, Webhook URL, Gmail 주소/앱 비밀번호를 하드코딩하지 않습니다.

**이번 STEP에서 하지 않는 것**

- `main.py` 생성 (STEP 19에서 진행)
- `crawler.py`, `openai_client.py` 생성 (아직 만들지 않음)
- Slack/Gmail 실제 재발송, OpenAI 재호출
- class, dataclass, service layer, dependency injection 같은 과도한 구조화 — 단순한 Python 모듈 + 함수만 사용합니다.

**Notebook과 src의 차이**

Notebook은 실험과 검증 기록을 남기는 곳이고, `src`는 검증이 끝난 코드를 재사용 가능한 Python 모듈로 관리하는 곳입니다. 그래서 STEP 17에서 만든 함수 정의 Cell은 지우지 않고 "검증 기록"으로 그대로 남겨둡니다.

**Notebook에서 src를 import하는 방법**

Notebook은 보통 `.../ax-job-agent/notebooks` 위치에서 실행되고, `src`는 그 한 단계 위(`.../ax-job-agent/src`)에 있습니다. `os.chdir()`로 작업 디렉터리를 바꾸는 대신, 프로젝트 루트 경로를 **한 번만** `sys.path`에 추가해서 `from src.analyzer import ...` 형태로 import합니다.

**성공 기준**

- `src/` 폴더와 3개 모듈(`analyzer.py`, `reporter.py`, `notifier.py`) + `__init__.py` 존재
- Notebook에서 import 성공
- `filter_related_jobs()` 결과가 기존과 동일 (5건/5건/True)
